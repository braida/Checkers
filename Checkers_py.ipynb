```python
"""Pions """
class Pieces:
    """Defini par
    color
    position (x, y)
    """
    """Constructeur class pieces
     """
    def __init__(self,color,x,y):
        self.color = color
        self.x = x
        self.y = y

    def __str__(self):
        return self.color + "-" +str(self.x) + "  " + str(self.y)

    def __repr__(self):
        return self.color + "-" +str(self.x) + "  " + str(self.y)
    
```

```python
"""Players """
class Players:
    """Defini par son
    name
    """
    """Constructeur class player
     """
    def __init__(self,name):
        self.name = name
        self.points = 0

    def get_name(self):
        return self.name

    
```

```python
import sys
from IPython.display import Markdown, display

def printmd(string):
    display(Markdown(string))
    

"""Board """
class Board:
    """Defini par
    largeur 
    list <player>
    list <pieces> black
    list <pieces> white
    """
    """Constructeur class board"""
    
            
    def __init__(self,width):
        self.width = width
        self.next_player = '' 
        self.lst_players = []
        self.lst_pieces_black = []
        self.lst_pieces_white = []

    def get_width(self):
        return self.width


    def get_piece_incell(self, x, y):  #NC
        for p0 in self.lst_pieces_black:
            if p0.x == x and p0.y == y:
                return p0
        for p1 in self.lst_pieces_white:
            if p1.x == x and p1.y == y:
                return p1 
       
    def draw(self):
        game = ''
        
        for j in range(self.width):
            for i in range(self.width):
                k = self.get_piece_incell(j, i)
                if k == None:
                    game = game + '* ' + '|'
                else: 
                    game = game + k.color + " " + '|'
            game = game + ' ' + str(j) +'\n' 
        #col_y = ''
        #for i in range(self.width):
         #   col_y = col_y + str(i) + '\n'
        column = ''
        for x in range(self.width):
            column = column + str(x) + '__'
        
        print(column)
        print(game)  
                
    def draw_line(self,num_column, is_odd):
        variable = ''
        for i in range(num_column):
            if i%2 == 0:
                if is_odd: 
                    variable = variable + '* ' 
                else:
                    variable = variable + 'x '
            else:
                if is_odd:
                    variable= variable + 'x '
                else: 
                    variable = variable + '* '
        print(variable)
                        
    def set_up_board(self):
        l = range(self.width) 
        for j in range(3):
            if j%2 == 0:
                for i in l[::2]:
                    black_piece = Pieces('B', j, i)
                    white_piece = Pieces('W', (self.width - 1)- j,i )
                    self.lst_pieces_black.append(black_piece)
                    self.lst_pieces_white.append(white_piece)
            else:
                for i in list(set(l) - set(l[::2])):
                    black_piece = Pieces('B', j, i)
                    white_piece = Pieces('W', (self.width - 1)- j,i )
                    self.lst_pieces_black.append(black_piece)
                    self.lst_pieces_white.append(white_piece)
            
            
           
        print(self.lst_pieces_black[:4]) 
        print(self.lst_pieces_black[4:8])
        print(self.lst_pieces_black[8:12], '\n') 
        
        print(self.lst_pieces_white[:4])  
        print(self.lst_pieces_white[4:8])
        print(self.lst_pieces_white[8:12])  
        

    def move_piece(self, x0, y0, x1, y1):
        if x1 >= self.width and y1 >= self.width and x1 < 0 and y1 < 0:
            print('Mouvement illegal : Position destination x outside board')
            return False

        if x0 >= self.width and y0 >= self.width and x0 < 0 and y0 < 0:
            print('Mouvement illegal : Position origin x outside board')
            return False
        if type(x0) != int or type(y0) != int or type(x1) != int or type(y1)!= int:
            print('Mouvement illegal : Input is not numeric')
            return False

        var_moved_piece = self.get_piece_incell(x0,y0)

        if var_moved_piece is None:
            print('Mouvement illegal : Piece not exist')
            return False

        if self.get_piece_incell(x1, y1) != None:
            print('Mouvement illegal : Cell not available')
            return False
        
        if var_moved_piece.color == 'B':
            if (x1 == x0+1 and y1 == y0+1) or (x1 == x0 + 1 and y1 == y0-1):
                var_moved_piece.x = x1
                var_moved_piece.y = y1
                return True
            elif (x1 == x0+2 and y1 == y0 + 2):
                var_eated_piece = self.get_piece_incell(x0+1, y0+1)
                if var_eated_piece != None and var_eated_piece.color == 'W':
                    var_moved_piece.x = x1
                    var_moved_piece.y = y1
                    
                    #remove var_eated_piece from list white
                    self.lst_pieces_white.remove(var_eated_piece)
                    return True
           
                else:
                    print('Mouvement illegal : Cell not available')
                    return False
            elif (x1== x0+2 and y1 == y0-2):
                var_eated_piece = self.get_piece_incell(x0+1, y0-1)
                if var_eated_piece != None and var_eated_piece.color == 'W':
                    var_moved_piece.x = x1
                    var_moved_piece.y = y1
                    #remove var_eated_piece from list white
                    self.lst_pieces_white.remove(var_eated_piece)
                    return True
                else:
                    print('Mouvement illegal : Cell not available')
                    return False
            else:
                print('Mouvement illegal')
                return False

        elif var_moved_piece.color == 'W':
            if (x1 == x0-1 and y1 == y0-1) or (x1 == x0 - 1 and y1 == y0+1):
                var_moved_piece.x = x1
                var_moved_piece.y = y1
                return True
            elif (x1 == x0-2 and y1 == y0 - 2):
                var_eated_piece = self.get_piece_incell(x0-1, y0-1)
                if var_eated_piece != None and var_eated_piece.color == 'B':
                    var_moved_piece.x = x1
                    var_moved_piece.y = y1
                    #remove var_eated_piece from list black
                    self.lst_pieces_black.remove(var_eated_piece)
                    return True
                else:
                    print('Mouvement illegal : Cell not available')
                    return False
            elif (x1== x0-2 and y1 == y0+2):
                var_eated_piece = self.get_piece_incell(x0-1, y0+1)
                if var_eated_piece != None and var_eated_piece.color == 'B':
                    var_moved_piece.x = x1
                    var_moved_piece.y = y1
                    #remove var_eated_piece from list black
                    self.lst_pieces_black.remove(var_eated_piece)
                    return True
                else:
                    print('Mouvement illegal : Cell not available')
                    return False
            else:
                print('Mouvement illegal')
                return False

    def is_game_over(self):
        if len(self.lst_pieces_white) == 0:
            return True
        elif len(self.lst_pieces_black) == 0:
            return True
        else: 
            return False
              
    def start(self):
        print('Player1 Enter your name')
        nom_joueur1 = input()  
        joueur1 = Players(nom_joueur1)
        

        print('Player2 Enter your name')
        nom_joueur2 = input() + '\n'
        joueur2 = Players(nom_joueur2) 

        #self.lst_players.append(joueur1)
        #self.lst_players.append(joueur2
        self.next_player = joueur1
        

        while not self.is_game_over():
            print('\n GAME ON')
            self.draw()
            print(self.next_player.name,'enter position of the current piece and it\'s destination (x0,y0,x1,y1):')
            #print(self.next_player.name, 'Now, let\'s PLAY: Enter position of the current piece and its new positions (x0,y0,x1,y1):'+ '\n')
            position = input()
            
            x0, y0, x1, y1 = position.split(',')
            is_move_piece = self.move_piece(int(x0),int(y0),int(x1),int(y1))

            if is_move_piece:
                if self.next_player.name == joueur1.name:
                    self.next_player = joueur2
                else:
                    self.next_player = joueur1
            else:
                print('')            

        if self.lst_pieces_black == []:
            print('{} Wins!'.format(joueur2))
        elif self.lst_pieces_white == []:
            print('{} Wins!'.format(joueur1))
        
```
